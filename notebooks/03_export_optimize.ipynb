# 03 — Export ONNX & Ottimizzazione TensorRT (Jetson)

Questo notebook va eseguito **sul Jetson Orin Nano** (non su Colab), dopo aver scaricato i modelli `.pt`.

```python
# 1. Export YOLO .pt → ONNX
from ultralytics import YOLO
model = YOLO('yolov8n_tennis.pt')
model.export(format='onnx', imgsz=640, simplify=True)
print('YOLO ONNX exported')
```

```python
# 2. Converti ONNX → TensorRT engine (FP16)
# Esegui da terminale sul Jetson:
# trtexec --onnx=yolov8n_tennis.onnx --saveEngine=yolov8n_tennis.engine --fp16 --workspace=4096
print('Run trtexec on Jetson terminal')
```
